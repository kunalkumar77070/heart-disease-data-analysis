This project performs exploratory data analysis (EDA) on the Cleveland Heart Disease dataset from the UCI Machine Learning Repository.

> **Note:** The original UCI target (`num`) is 0–4. For a simple binary analysis, this notebook converts it to `target`: `0 = no disease`, `1 = disease present`.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

## 1. Load the Dataset

The data is downloaded from the UCI Machine Learning Repository. The original Cleveland dataset contains 303 records and 13 input features plus the diagnosis field.


```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"

columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs",
    "restecg", "thalach", "exang", "oldpeak", "slope",
    "ca", "thal", "num"
]

df = pd.read_csv(url, names=columns, na_values="?")

# Convert the original 0–4 diagnosis into a binary target:
# 0 = no disease, 1 = disease present
df["target"] = (df["num"] > 0).astype(int)
df.drop(columns=["num"], inplace=True)

print("Dataset shape:", df.shape)
df.head()
```

## 2. Basic Dataset Information

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

## 9. Number of Patients

## 10. Target Class Distribution

```python
print(df["target"].value_counts().sort_index())

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="target")
plt.title("Heart Disease Target Distribution")
plt.xlabel("Target (0 = No Disease, 1 = Disease)")
plt.ylabel("Number of Patients")
plt.show()
```

## 11. Age Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="age", bins=15, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

## 12. Cholesterol and Resting Blood Pressure Boxplots

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

sns.boxplot(y=df["chol"], ax=axes[0])
axes[0].set_title("Cholesterol")

sns.boxplot(y=df["trestbps"], ax=axes[1])
axes[1].set_title("Resting Blood Pressure")

plt.tight_layout()
plt.show()
```

## 13. Chest Pain Type Distribution

```python
plt.figure(figsize=(7,4))
sns.countplot(data=df, x="cp")
plt.title("Chest Pain Type Distribution")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")
plt.show()
```

## 14. Correlation Heatmap

```python
plt.figure(figsize=(11,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 15. Key Observations

- The dataset can be inspected for missing values and duplicate records before analysis.
- Age, cholesterol, resting blood pressure and maximum heart rate can be explored using summary statistics and visualizations.
- The target variable is converted into a binary classification label for simple EDA.
- The correlation heatmap helps identify linear relationships between numerical variables.
